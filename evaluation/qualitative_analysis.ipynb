```python
def load_codings(path, model, judge):
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    rows = []

    for dialogue in data["dialogues"]:

        # Ignore malformed/non-dialogue entries
        if not isinstance(dialogue, dict):
            continue

        if "global_id" not in dialogue:
            continue

        global_id = dialogue["global_id"]

        for coding in dialogue.get("codings", []):

            if not isinstance(coding, dict):
                continue

            rows.append({
                "model": model,
                "judge": judge,
                "global_id": global_id,
                "speaker": coding.get("speaker"),
                "category": coding.get("category"),
                "indicator": coding.get("indicator"),
                "text_segment": coding.get("text_segment", ""),
                "reasoning": coding.get("reasoning", "")
            })

    return pd.DataFrame(rows)
```

## Looking for Example Candidates of Judge Agreements and Disagreements

```python
import pandas as pd

# 1. define status: agreement or disagreement
# did all three judges code the same dialogue?
def determine_status_strict(row):
    judge_cols = ["Judge 1 (Qwen3-32B)", "Judge 2 (Qwen3-30B)", "Judge 3 (Mistral-Small-24B)"]
    values = []
    
    for col in judge_cols:
        val = row[col]
        if isinstance(val, list) and len(val) > 0:
            values.append(val[0])
        else:
            # if empty, not eligible for example
            return "Missing Coding"
            
    # if set is == 1 -> agreement
    if len(set(values)) == 1:
        return "Clear Agreement"
    else:
        # if set is > 1 disagreement
        return "Disagreement"

comparison["status"] = comparison.apply(determine_status_strict, axis=1)

# 2. get text, category and reasoning
raw_data = (
    all_codings.groupby(["model", "global_id", "speaker", "dimension", "judge"])[
        ["text_segment", "category", "reasoning"]
    ]
    # filter empty codings or join with "|" in case theres more than one
    .agg(lambda x: " | ".join(set(str(v) for v in x if pd.notna(v) and str(v).strip() != "")))
    .reset_index()
)

data_pivot = raw_data.pivot_table(
    index=["model", "global_id", "speaker", "dimension"],
    columns="judge",
    values=["text_segment", "category", "reasoning"],
    aggfunc="first"
)

# flatten data
data_pivot.columns = [f"{col[0]}_{col[1]}" for col in data_pivot.columns]
data_pivot = data_pivot.reset_index()

# merge with comparison dataframe
qualitative_analysis_df = comparison.merge(
    data_pivot, 
    on=["model", "global_id", "speaker", "dimension"], 
    how="left"
)

# 3. extract examples (randomised & model-diverse)

# printing function
def print_judge_info(judge_id, indicator, category, text, reasoning):
    print(f"  {judge_id} (Indicator: {indicator})")
    print(f"    Category : {category}")
    print(f"    Text     : {text}")
    print(f"    Reasoning: {reasoning}")


print("--- CANDIDATES CLEAR AGREEMENT ---")
# mix all Agreement-cases (frac=1), throw duplicate at 'model' out
# so that each model max 1x in examples, extract 5 examples
clear_cases = (
    qualitative_analysis_df[qualitative_analysis_df["status"] == "Clear Agreement"]
    .sample(frac=1) 
    .drop_duplicates(subset=["model"])
    .head(5)
)

for _, row in clear_cases.iterrows():
    print(f"\nModell: {row['model']} | Dialog: {row['global_id']} | Speaker: {row['speaker']} | Dimension: {row['dimension']}")
    
    print_judge_info("J1", row['Judge 1 (Qwen3-32B)'], row['category_Judge 1 (Qwen3-32B)'], row['text_segment_Judge 1 (Qwen3-32B)'], row['reasoning_Judge 1 (Qwen3-32B)'])
    print_judge_info("J2", row['Judge 2 (Qwen3-30B)'], row['category_Judge 2 (Qwen3-30B)'], row['text_segment_Judge 2 (Qwen3-30B)'], row['reasoning_Judge 2 (Qwen3-30B)'])
    print_judge_info("J3", row['Judge 3 (Mistral-Small-24B)'], row['category_Judge 3 (Mistral-Small-24B)'], row['text_segment_Judge 3 (Mistral-Small-24B)'], row['reasoning_Judge 3 (Mistral-Small-24B)'])
    print("-" * 80)


print("\n--- CANDIDATES DISAGREEMENT ---")
# Gleiche Logik für die Disagreements
disagreement_cases = (
    qualitative_analysis_df[qualitative_analysis_df["status"] == "Disagreement"]
    .sample(frac=1)
    .drop_duplicates(subset=["model"])
    .head(5)
)

for _, row in disagreement_cases.iterrows():
    print(f"\nModel: {row['model']} | Dialog: {row['global_id']} | Speaker: {row['speaker']} | Dimension: {row['dimension']}")
    
    print_judge_info("J1", row['Judge 1 (Qwen3-32B)'], row['category_Judge 1 (Qwen3-32B)'], row['text_segment_Judge 1 (Qwen3-32B)'], row['reasoning_Judge 1 (Qwen3-32B)'])
    print_judge_info("J2", row['Judge 2 (Qwen3-30B)'], row['category_Judge 2 (Qwen3-30B)'], row['text_segment_Judge 2 (Qwen3-30B)'], row['reasoning_Judge 2 (Qwen3-30B)'])
    print_judge_info("J3", row['Judge 3 (Mistral-Small-24B)'], row['category_Judge 3 (Mistral-Small-24B)'], row['text_segment_Judge 3 (Mistral-Small-24B)'], row['reasoning_Judge 3 (Mistral-Small-24B)'])
    print("-" * 80)
```

```python
# 4. Examples for exactly the same text segment 

def is_exact_text_match(row):
    t1 = str(row['text_segment_Judge 1 (Qwen3-32B)']).strip() if pd.notna(row['text_segment_Judge 1 (Qwen3-32B)']) else ""
    t2 = str(row['text_segment_Judge 2 (Qwen3-30B)']).strip() if pd.notna(row['text_segment_Judge 2 (Qwen3-30B)']) else ""
    t3 = str(row['text_segment_Judge 3 (Mistral-Small-24B)']).strip() if pd.notna(row['text_segment_Judge 3 (Mistral-Small-24B)']) else ""
    
    # test if text is the same for all three judges
    if t1 and t2 and t3 and (t1 == t2 == t3):
        return True
    return False

# new dataframe with exact text segment matches
exact_match_df = qualitative_analysis_df[qualitative_analysis_df.apply(is_exact_text_match, axis=1)]

print("--- CANDIDATES FOR EXACT TEXT SEGMENT MATCHES ---")

# mix and filter for models
exact_cases = (
    exact_match_df
    .sample(frac=1)
    .drop_duplicates(subset=["model"])
    .head(5)
)

for _, row in exact_cases.iterrows():
    print(f"\nModell: {row['model']} | Dialog: {row['global_id']} | Speaker: {row['speaker']} | Dimension: {row['dimension']}")
    print(f"Extrahierter Text (Identisch bei allen): '{row['text_segment_Judge 1 (Qwen3-32B)']}'\n")

    # Inidicators, reasoning and category per judge
    print(f"  J1 (Qwen3-32B)        | Indicator: {row['Judge 1 (Qwen3-32B)']}")
    print(f"    Category : {row['category_Judge 1 (Qwen3-32B)']}")
    print(f"    Reasoning: {row['reasoning_Judge 1 (Qwen3-32B)']}\n")
    
    print(f"  J2 (Qwen3-30B)        | Indicator: {row['Judge 2 (Qwen3-30B)']}")
    print(f"    Category : {row['category_Judge 2 (Qwen3-30B)']}")
    print(f"    Reasoning: {row['reasoning_Judge 2 (Qwen3-30B)']}\n")
    
    print(f"  J3 (Mistral-Small-24B)| Indicator: {row['Judge 3 (Mistral-Small-24B)']}")
    print(f"    Category : {row['category_Judge 3 (Mistral-Small-24B)']}")
    print(f"    Reasoning: {row['reasoning_Judge 3 (Mistral-Small-24B)']}")
    print("-" * 80)
```

```python
import numpy as np
import krippendorff

# take exact text segment df
# exact_match_df = qualitative_analysis_df[qualitative_analysis_df.apply(is_exact_text_match, axis=1)]

dimensions = ["E/I", "N/S", "T/F", "J/P"]
judge_cols = ["Judge 1 (Qwen3-32B)", "Judge 2 (Qwen3-30B)", "Judge 3 (Mistral-Small-24B)"]

print("--- Krippendorff's Alpha for Exact Text Segment Matches ---")

for dim in dimensions:
    # 1. filter for current mbti dimension
    df_dim = exact_match_df[exact_match_df["dimension"] == dim]
    
    if len(df_dim) == 0:
        print(f"{dim}: No exact matches found.")
        continue
        
    # 2. Rater-Matrix (Rows = Judges, Cols = text segments)
    matrix = []
    for judge in judge_cols:
        judge_ratings = []
        for val in df_dim[judge]:
            # first element of list (e.g 'E' out of ['E'])
            if isinstance(val, list) and len(val) > 0:
                judge_ratings.append(val[0])
            else:
                judge_ratings.append(np.nan)
        matrix.append(judge_ratings)
        
    # 3. Strings into numeric values for krippendorff.alpha (np.nan stays)
    # Find all unique valid labels (z.B. 'E', 'I')
    unique_labels = set(x for row in matrix for x in row if pd.notna(x))
    label_to_id = {label: i for i, label in enumerate(sorted(list(unique_labels)))}
    
    num_matrix = []
    for row in matrix:
        num_matrix.append([label_to_id[x] if pd.notna(x) else np.nan for x in row])
        
    # 4. calculate Krippendorff's Alpha
    try:
        alpha = krippendorff.alpha(reliability_data=num_matrix, level_of_measurement='nominal')
        print(f"{dim}: α = {alpha:.3f} (based on {len(df_dim)} identical text segments)")
    except Exception as e:
        print(f"{dim}: Couldn't calculate ({e})")
```
