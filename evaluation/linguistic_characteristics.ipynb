# Linguistic Characteristics

Calculating English-rate, distinct-n and especially for the qwen models the amount of chines characters.

```python
import json
import pandas as pd
from langdetect import detect


def detect_language(text):
    try:
        return detect(text)
    except:
        return "unknown"


def load_dialogues(filepath):
    with open(filepath, "r", encoding="utf-8") as f:
        return json.load(f)


def extract_utterances(raw_data, model_name):
    rows = []

    for dialogue in raw_data:
        for utt in dialogue["Utterances_A"]:
            rows.append({
                "model": model_name,
                "speaker": "A",
                "global_id": dialogue.get("global_id", dialogue.get("Run")),
                "text": utt
            })

        for utt in dialogue["Utterances_B"]:
            rows.append({
                "model": model_name,
                "speaker": "B",
                "global_id": dialogue.get("global_id", dialogue.get("Run")),
                "text": utt
            })

    return pd.DataFrame(rows)


#JSON files with the llm-as-a-judge evals
files = {
    "qwen_base": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_qwen_base.json",
    "qwen_lora1": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_qwen_lora1_correct.json",
    "qwen_lora2": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_qwen_lora2.json",
    "mistral_base": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_ministral_base.json",
    "mistral_lora1": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_ministral_lora1.json",
    "mistral_lora2": "C:/Users/Tim/Projects/MA/data/llmasjudge_evals_pub/json_ground_truth_dias/dialogues_ministral_lora2.json"
}


dfs = {}

for model_name, filepath in files.items():
    raw_data = load_dialogues(filepath)
    dfs[model_name] = extract_utterances(raw_data, model_name)


for model_name, df in dfs.items():
    df["language"] = df["text"].apply(detect_language)

    print(f"\n--- {model_name} ---")
    print(df["language"].value_counts())
```

```python
df_all = pd.concat(dfs.values(), ignore_index=True)

df_all["language"] = df_all["text"].apply(detect_language)

language_table = pd.crosstab(
    df_all["model"],
    df_all["language"]
)

print(language_table)
```

## Distinct-n

```python
def distinct_n(texts, n):
    """
    Calculates Distinc-n
    Distinct-n = Number unique n-grams / Number of all n-grams
    """
    ngrams = []

    for text in texts:
        tokens = text.lower().split()

        if len(tokens) >= n:
            ngrams.extend(
                tuple(tokens[i:i+n])
                for i in range(len(tokens) - n + 1)
            )

    if not ngrams:
        return 0

    return len(set(ngrams)) / len(ngrams)
```

```python
results = []

for model_name, df in dfs.items():

    distinct_1 = distinct_n(df["text"], 1)
    distinct_2 = distinct_n(df["text"], 2)

    results.append({
        "model": model_name,
        "distinct_1": distinct_1,
        "distinct_2": distinct_2
    })

distinct_results = pd.DataFrame(results)

print(distinct_results)
```

```python
results = []

for model_name, df in dfs.items():

    for speaker in ["A", "B"]:

        speaker_df = df[df["speaker"] == speaker]

        distinct_1 = distinct_n(speaker_df["text"], 1)
        distinct_2 = distinct_n(speaker_df["text"], 2)

        results.append({
            "model": model_name,
            "speaker": speaker,
            "distinct_1": distinct_1,
            "distinct_2": distinct_2
        })

distinct_speaker_results = pd.DataFrame(results)

print(distinct_speaker_results) 
```
